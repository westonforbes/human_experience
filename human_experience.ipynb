### Define points of interest in their native units.

```python
# X axis values.
planck_length = 1.616255e-35  # The smallest possible length, in meters.
smallest_human_scale_length = .1 # The smallest length people regularly interact with, in millimeters.
largest_human_scale_length = 12756 # The farthest the average person can travel, the diameter of the Earth, in kilometers.
observable_universe_diameter = 93e9  # The largest observable length, in light-years.

# Y axis values.
planck_time = 5.391247e-44  # The smallest possible unit of time, in seconds
smallest_human_scale_time = 30 # The rough temporal resolution of human perception, in milliseconds.
largest_human_scale_time = 100 # A long human lifetime, in years.
age_of_universe = 13.8e9 # The age of the universe, in years.
```

### Convert points of interest to common units.
Additionally, relabel from smallest to largest by axis.

```python
x1 = planck_length # Already in meters.
x2 = smallest_human_scale_length * 1e-3  # Convert to meters.
x3 = largest_human_scale_length * 1e3  # Convert to meters.
x4 = observable_universe_diameter * 9.461e15  # Convert to meters.

y1 = planck_time  # Already in seconds.
y2 = smallest_human_scale_time * 1e-3  # Convert to seconds.
y3 = largest_human_scale_time * 365.25 * 24 * 3600  # Convert to seconds.
y4 = age_of_universe * 365.25 * 24 * 3600  # Convert to seconds.
```

### Plot

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 6))
plt.xscale('log')
plt.yscale('log')

# Plot rectangle.
plt.plot([x2, x3, x3, x2, x2], [y2, y2, y3, y3, y2], color='red')

# Set axis limits.
plt.xlim(x1, x4)
plt.ylim(y1, y4)

plt.xlabel('Length (meters, log scale)')
plt.ylabel('Time (seconds, log scale)')
plt.title('Human Perception of Reality')
plt.show()
```
